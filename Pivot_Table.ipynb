```python
import pandas as pd
from pydataset import data
df = data("mtcars")
df.head()
```

```python
pd.pivot_table(df,values ="mpg",index="cyl") # Avg MPG by num of cyls
```

```python
pd.pivot_table(df,values="hp",index="cyl",aggfunc="sum") # total hp by cyls
```

```python
pd.pivot_table(df,values="mpg",index="cyl",aggfunc="count") # Num of Cars in each Cyls group
```

```python
pd.pivot_table(df,values=["mpg","hp"],index="cyl") # Avg MPG & HP
```

```python
pd.pivot_table(df,values="mpg",index="cyl",aggfunc=["mean","max","min"]) # Multiple Aggregations
```

```python
pd.pivot_table(df,values="mpg",index="cyl",columns="am",aggfunc="mean") # Two Dimensional Pivot. Avg MPG by Cyl & Am
```

```python
pd.pivot_table(df,values="hp",index="gear",columns="am",aggfunc="mean") # Avg Hp
```

```python
pd.pivot_table(df,values="mpg",index=["cyl","gear"]) # Multiple Indexs
```

```python
pd.pivot_table(df,values=["mpg","hp"],index=["cyl","gear"]) # Multiple Values & Multiple Indexs
```

```python
pd.pivot_table(df,values="mpg",index="cyl")
```

```python
pd.pivot_table(df,values="mpg",index="cyl",columns="gear")
```

# Fill Missing Values

```python
pd.pivot_table(df,values="mpg",index="cyl",columns="gear",fill_value=0) # Fill missing values
```

```python
pd.pivot_table(df,values="mpg",index="cyl",columns="gear",fill_value=555)
```

# Grand Total(Margins)

```python
pd.pivot_table(df,values="mpg",index="cyl",aggfunc="mean",margins=True) # All rows shows the overall Avg MPG
```

```python
pd.pivot_table(df,values="mpg",index="cyl",columns="am",margins=True) # Grand total with rows & columns
```

```python
pd.pivot_table(df,values="wt",index="gear",aggfunc=["mean","sum","max","min"])
```

```python
pd.pivot_table(df,values=["mpg","hp","wt"],index="cyl",columns="am",aggfunc="mean")
```

```python
pd.pivot_table(df,values=["mpg","hp","wt"],index="cyl",columns="am",aggfunc="mean",fill_value=0)
```

# Interview Questions

# 1. Average MPG by Cylinders

```python
pd.pivot_table(df,values="mpg",index="cyl",aggfunc="mean")
```

# 2. Average MPG by Cylinders & Transmission

```python
pd.pivot_table(df,values="mpg",index="cyl",columns="am",aggfunc="mean")
```

# 3. Total Horsepower by Gear

```python
pd.pivot_table(df,values="hp",index="gear",aggfunc="sum")
```

# 4. Count Cars By Transmissions

```python
pd.pivot_table(df,values="mpg",index="am",aggfunc="count")
```

# 5. Average Weight By Gear

```python
pd.pivot_table(df,values="wt",index="gear",aggfunc="mean")
```

# 6. Compare Manual Vs Automatic Cars

```python
pd.pivot_table(df,values=["mpg","hp","wt"],index="am",aggfunc="mean")
```

# 7. Multiple Aggregations(Mean,Max,Min)

```python
pd.pivot_table(df,values="mpg",index="cyl",aggfunc=["mean","max","min"])
```

# 8. Add Grand Totals Using Margins=True

```python
pd.pivot_table(df,values="mpg",index="cyl",aggfunc="mean",margins=True)
```

# 9. Replace Missing Values Using Fill_Value

```python
pd.pivot_table(df,values="mpg",index="cyl",columns="gear",fill_value=0)
```

# 10. Multiple Rows & Columns Groupings

```python
pd.pivot_table(df,values=["mpg","hp"],index=["cyl","gear"],columns="am",aggfunc="mean",fill_value=0,margins=True)
```
