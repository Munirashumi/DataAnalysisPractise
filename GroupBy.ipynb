```python
import pandas as pd
from pydataset import data
df = data("mtcars")
df.head()
```

# 1. Group by one Column

```python
df.groupby("cyl")["mpg"].mean() # Avg MPG by Cylinder
```

# 2. Count Cars in each cylinder

```python
df.groupby("cyl").size()
```

```python
df.groupby("cyl").count()
```

```python
df.groupby("cyl").value_counts()
```

# 3. Total Horsepower by Cylinder

```python
df.groupby("cyl")["hp"].sum()
```

# 4. Minimum MPG by Cylinder

```python
df.groupby("cyl")["mpg"].min()
```

# 5. Maximum MPG by Cylinder

```python
df.groupby("cyl")["mpg"].max()
```

# 6. Multiple Aggregations

```python
df.groupby("cyl")["mpg"].agg(["mean","max","min","count"])
```

# 7. Group by Multiple Columns

```python
df.groupby(["cyl","am"])["mpg"].mean()
```

# 8. Multiple Columns Aggregation

```python
df.groupby("gear")[["mpg","hp"]].mean()
```

# 9. Mean of All Numeric Columns

```python
df.groupby("gear").mean(numeric_only=True)
```

```python
df.groupby("gear")["mpg"].mean(numeric_only=True)
```

# 10. Aggregate Different Functions

```python
df.groupby("mpg").agg({"gear":"mean","hp":"min","cyl":"max"})
```

# 11. Rename Aggregate Columns

```python
df.groupby("gear").agg(Average_MPG=("mpg","mean"),Total_HP=("hp","sum"),Max_Weight=("wt","max"))
```

# 12. Find Highest Hp Car in Each Cylinder

```python
df.groupby("cyl")["hp"].max()
```

# 13. Find Lowest Wt Car in Each Cylinder¶

```python
df.groupby("cyl")["wt"].min()
```

# 14. Count Manual Vs Automatic

```python
df.groupby("am").size()
```

# 15. Average Wt by Gear

```python
df.groupby("gear")["wt"].mean()
```

# 16. Average Quarter Mile time

```python
df.groupby("gear")["qsec"].mean()
```

# 17. Sort Group Results

```python
df.groupby("cyl")["mpg"].mean().sort_values(ascending=False)
```

# 18. Multiple Statistics

```python
df.groupby("gear")["hp"].describe()
```

```python
df.groupby("cyl")["hp"].describe()
```

# 19. Custom Aggregation

```python
df.groupby("cyl")["mpg"].agg(lambda x : x.max()-x.min()) # Range= Max-min
```

# 20. Group BY AND Reset Index

```python
df.groupby("gear")["mpg"].mean().reset_index()
```

# 21. Named Aggregation

```python
df.groupby("cyl").agg(Avg_Mpg=("mpg","mean"), Avg_Hp=("hp","mean"),Count=("wt","count"))
```

# 22. Filter Groups

```python
df.groupby("cyl").filter(lambda x :len(x)>10)
```

# 22.Transform

```python
df["Avg_Mpg"]= df.groupby("cyl")["mpg"].transform("mean")
df
```

# 24. Rank Within Each Cylinder

```python
df["HP_Rank"]= df.groupby("cyl")["hp"].rank(ascending=False)
df
```

# 25. Highest Horsepower Car per Cylinder

```python
df.loc[df.groupby("cyl")["hp"].idxmax()]
```

# 26. Lowest MPG Car per Cylinder

```python
df.loc[df.groupby("cyl")["mpg"].idxmin()]
```

# 27. Apply Custom Function

```python
def mpg_range(group) : return group.max()-group.min()
df.groupby("cyl")["mpg"].apply(mpg_range)
```

```python
def mpg_range(x) : return x.max()-x.min()
df.groupby("cyl")["mpg"].apply(mpg_range)
```

# 28. Percentage of Total Horsepower

```python
df["Hp Percentage"]= (df["hp"] / df["hp"].sum())*100
df
```

# 29.Percentage Within Each Cylinder

```python
df["Cylinder_Hp%"]=(df["hp"] / df.groupby("cyl")["hp"].transform("sum"))*100
df
```

# 36. Groupby + cumcount()

```python
df["Row_No"]=df.groupby("cyl")["hp"].cumcount() # Assign row numbers within each cylinder group
df
```

# 37. Groupby + Cumsum()

```python
df["Running_HP"]= df.groupby("cyl")["hp"].cumsum() # Running total horsepower within each cylinder
df
```

# 38. Groupby + Shift()

```python
df["Previous_Hp"]= df.groupby("cyl")["hp"].shift()
df
```

# 39. Groupby + Diff()

```python
df["HP_Diff"]=df.groupby("cyl")["hp"].diff() # Difference in horsepower from previous row
df
```

# 40. Groupby + Rolling()

```python
df["Rolling_MPG"]= (df.groupby("cyl")["mpg"].rolling(2).mean().reset_index(level=0,drop=True))
df
```
