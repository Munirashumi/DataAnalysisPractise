# 14. Merge()

```python
cars = pd.DataFrame({"CarID": [1, 2, 3],
                     "Model": ["Mazda RX4", "Datsun 710", "Valiant"]})

price = pd.DataFrame({"CarID": [1, 2, 3],
                      "Price": [25000, 18000, 22000]})

merged = pd.merge(cars, price, on="CarID")

print(merged)
```

```python
pd.merge(cars, price, on="CarID")
```

```python
pd.merge(cars, price, on="CarID",how="right")
```

```python
pd.merge(cars, price, on="CarID",how="left")
```

```python
pd.merge(cars, price, on="CarID",how="inner")
```

```python
pd.merge(cars, price, on="CarID",how="outer")
```
